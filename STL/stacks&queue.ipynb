## Stack

```python

# ************************************
# STACK -> FILO (FIRST IN LAST OUT)
# ************************************
from collections import deque

# USING COLLECTIONS DEQUE
stack = deque()

stack.append(10)
stack.append(11)
stack.append(12)

print(stack[-1]) # TOP
print(len(stack)) # LENGTH

print(stack.pop())
print(stack.pop())
print(stack.pop())
```

```python
# 2. USING PYTHON'S LIST
stack = []

stack.append(10)
stack.append(11)
stack.append(12)

print(stack)
print(len(stack))

print(stack.pop())
print(stack.pop())
print(stack.pop())
```

## Queue

```python
# ************************************
# QUEUE -> FIFO (FIRST IN FIRST OUT)
# ************************************
from collections import deque

queue = deque()

queue.append(1)
queue.append(2)
queue.append(3)

print(len(queue))
print(queue[-1]) # front
print(queue[0]) # back

print(queue.popleft())
print(queue.popleft())
print(queue.popleft())
```

## Priority queue

```python
# ************************************************************************
# PRIORITY QUEUE (MIN HEAP)
# The element with highest priority is removed first
# ************************************************************************
import heapq

pq = []

# BY DEFAULT THE IT IS MIN HEAP SO SMALL ELEMENTS CONES FIRST
heapq.heappush(pq, 5)
heapq.heappush(pq, 1)
heapq.heappush(pq, 3)
heapq.heappush(pq, 1)

print(pq) # It is not sorted

print(heapq.heappop(pq))
print(heapq.heappop(pq))
print(heapq.heappop(pq))
```

```python
# ********************************************************************************************
# PRIORITY QUEUE (MAX HEAP TRICK)
# Python does not have built in max heap, so the trick is to store -ve numbers in the heap.
# ********************************************************************************************
import heapq

pq = []

heapq.heappush(pq, -1)
heapq.heappush(pq, -5)
heapq.heappush(pq, -2)
heapq.heappush(pq, 0)

print(pq)

print(heapq.heappop(pq))
print(heapq.heappop(pq))
print(heapq.heappop(pq))
print(heapq.heappop(pq))
```

```python
# ********************************************************************************************
# PRIORITY QUEUE WITH TUPLES
# Python does not have built in max heap, so the trick is to store -ve numbers in the heap.
# ********************************************************************************************

import heapq

pq = []

# (PRIORITY, VALUE)
# First it compares priority, then value if needed.
heapq.heappush(pq, (2, "task2"))
heapq.heappush(pq, (1, "task1"))
heapq.heappush(pq, (3, "task3"))

print(heapq.heappop(pq))
print(heapq.heappop(pq))
print(heapq.heappop(pq))

# *** imp ***
# If these values are not comparable
# heapq.heappush(pq, (2, {'name': 'Tejasva'}))
# heapq.heappush(pq, (2, {'name': 'Aarjav'}))
# heapq.heappush(pq, (2, {'name': 'Aaryan'}))

# So there should be a tie breaker so we can do it like this:
i = 0
heapq.heappush(pq, (2, i, {'name': 'Tejasva'}))
i = 20
heapq.heappush(pq, (2, i, {'name': 'Aarjav'}))
i = 3
heapq.heappush(pq, (2, i, {'name': 'Aaryan'}))

print(heapq.heappop(pq))
print(heapq.heappop(pq))
print(heapq.heappop(pq))
```

## Sets

```python
s = set()

s.add(10)
s.add(11)
print(s)

print(s.remove(10)) # RAISES ERROR IF VALUE NOT FOUND
print(s)

print(s.discard(13)) # DO NOT RAISES ERROR IF VALUE NOT FOUND
print(s)

s.add(10)
s.add(11)
s.add(13)
print(s.clear()) # clears the entire set
print(s)
```
